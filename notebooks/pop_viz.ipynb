```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import numpy as np
import seaborn as sns
```

# Population level TR vizuals

## Data inputs

```python
novelty_tsv_path = r"first_500_INS.novelty.tsv"

nv_df = pd.read_csv(novelty_tsv_path, sep="\t")
not_known_df = nv_df[nv_df["novelty"] != "known"]
known_df = nv_df[nv_df["novelty"] == "known"]
```

## Comparisons

Comparing ditributions of TR stats between novelty types:

### Motif Length comparisons

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(
    data=nv_df, 
    x='motif_length', 
    hue='novelty', 
    fill=True,       
    alpha=0.3,       
    linewidth=2
)
plt.title('Motif Length Density: Novel vs. Known')
plt.xlabel('Motif Length')
plt.ylabel('Density')
plt.tight_layout()
plt.show()
```

```python
sns.displot(
    data=nv_df, 
    x='motif_length', 
    col='novelty',     
    kde=True,
    bins=30,
    height=4, 
    aspect=1.2,
    stat="density"      
)
plt.subplots_adjust(top=0.85)
plt.suptitle('Insertion Length Distribution by Novelty')
plt.show()
```

### Repeat length

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(
    data=nv_df, 
    x='rep_length', 
    hue='novelty', 
    fill=True,       
    alpha=0.3,       
    linewidth=2
)
plt.title('Repeat Length Density: Novel vs. Known')
plt.xlabel('Repeat Length')
plt.ylabel('Density')
plt.tight_layout()
plt.show()
```

```python
sns.displot(
    data=nv_df, 
    x='rep_length', 
    col='novelty',     
    kde=True,
    bins=30,
    height=4, 
    aspect=1.2,
    stat="density"      
)
plt.subplots_adjust(top=0.85)
plt.suptitle('Repeat Length Distribution by Novelty')
plt.show()
```

### Purity

```python
plt.figure(figsize=(10, 6))
sns.kdeplot(
    data=nv_df, 
    x='purity', 
    hue='novelty', 
    fill=True,       
    alpha=0.3,       
    linewidth=2
)
plt.title('Purity Density: Novel vs. Known')
plt.xlabel('Purity')
plt.ylabel('Density')
plt.tight_layout()
plt.show()
```

```python
sns.displot(
    data=nv_df, 
    x='purity', 
    col='novelty',     
    kde=True,
    bins=30,
    height=4, 
    aspect=1.2     
)
plt.subplots_adjust(top=0.85)
plt.suptitle('Purity Distribution by Novelty')
plt.show()
```

## Full Population Visualizations

```python
# 1. Cap any length greater than 30 at 30
length_clipped = nv_df["motif_length"].clip(upper=30)

# 2. Plot from 0 to 30 with a fixed bin width
ax = sns.histplot(x=length_clipped, binrange=(0, 30), discrete=True)

# 3. Lock x-axis boundaries exactly between 0 and 30
ax.set_xlim(0, 30)

# 4. Format ticks so 30 renders as "30+"
ticks = [t for t in ax.get_xticks() if 0 <= t <= 30]
ax.set_xticks(ticks)
ax.set_xticklabels([f"{int(t)}+" if t == 30 else f"{int(t)}" for t in ticks])
ax.set_title("Cross-Sample Motif Length Distribution")
```

```python
upper = 3000

# 1. Cap any length greater than 30 at 30
length_clipped = nv_df["rep_length"].clip(upper=upper)

# 2. Plot from 0 to 30 with a fixed bin width
ax = sns.histplot(x=length_clipped, binrange=(0, upper), binwidth=50)

# 3. Lock x-axis boundaries exactly between 0 and 30
ax.set_xlim(0, upper)

# 4. Format ticks so 30 renders as "30+"
ticks = [t for t in ax.get_xticks() if 0 <= t <= upper]
ax.set_xticks(ticks)
ax.set_xticklabels([f"{int(t)}+" if t == upper else f"{int(t)}" for t in ticks])
ax.set_title("Cross-Sample Repeat Length Distribution")
```
